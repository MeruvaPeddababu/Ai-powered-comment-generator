```python
import requests
import json
import gradio as gr
from datetime import datetime
from collections import deque

# Configuration
OPENROUTER_API_URL = "https://openrouter.ai/api/v1/chat/completions"
HEADERS = {
    "HTTP-Referer": "https://colab.research.google.com/",
    "X-Title": "Healthcare AI Comment Generator"
}

class CommentGenerator:
    def __init__(self):
        self.history = deque(maxlen=10)
        self.current_comments = []

    def analyze_post(self, post_content: str, api_key: str) -> dict:
        """Extract key points from post"""
        headers = {**HEADERS, "Authorization": f"Bearer {api_key}"}

        prompt = f"""Analyze this healthcare AI LinkedIn post and extract:
1. 3 key discussion points
2. 2 areas where expert perspective adds value

Return as JSON:
{{
    "key_points": ["...", "...", "..."],
    "value_adds": ["...", "..."]
}}

Post:
{post_content}"""

        try:
            response = requests.post(
                OPENROUTER_API_URL,
                headers=headers,
                json={
                    "model": "meta-llama/llama-3-70b-instruct",
                    "messages": [{"role": "user", "content": prompt}],
                    "response_format": {"type": "json_object"},
                    "temperature": 0.3
                },
                timeout=30
            )
            return json.loads(response.json()["choices"][0]["message"]["content"])
        except Exception as e:
            print(f"Analysis error: {str(e)}")
            return None

    def generate_comments_with_emojis(self, post_content: str, client_bio: str, api_key: str) -> list:
        """Generate 2-3 professional comments with integrated emojis"""
        analysis = self.analyze_post(post_content, api_key)
        if not analysis:
            return ["[ERROR] Analysis failed"]

        headers = {**HEADERS, "Authorization": f"Bearer {api_key}"}

        prompt = f"""Generate 3 professional LinkedIn comments (2-3 sentences each) that:
1. Reference: {analysis['key_points'][0]}
2. Incorporate: {client_bio}
3. Add unique value about: {analysis['value_adds'][0]}
4. Position the author as a thought leader
5. Start with 1 emoji and end with 1-2 emojis

Post Content:
{post_content}

Return as a JSON list:
{{
    "comments": [
        "comment 1 with emojis",
        "comment 2 with emojis",
        "comment 3 with emojis"
    ]
}}"""

        try:
            response = requests.post(
                OPENROUTER_API_URL,
                headers=headers,
                json={
                    "model": "anthropic/claude-3-sonnet",
                    "messages": [{"role": "user", "content": prompt}],
                    "response_format": {"type": "json_object"},
                    "temperature": 0.7,
                    "max_tokens": 500
                },
                timeout=40
            )

            result = json.loads(response.json()["choices"][0]["message"]["content"])
            comments = result.get("comments", [])
            self.current_comments = comments

            # Add to history
            timestamp = datetime.now().strftime("%m/%d %H:%M")
            for comment in comments[:3]:
                self.history.append({
                    "time": timestamp,
                    "post": post_content[:50] + ("..." if len(post_content) > 50 else ""),
                    "comment": comment
                })

            return comments[:3]

        except Exception as e:
            print(f"Generation error: {str(e)}")
            return ["[ERROR] Comment generation failed"]

# Initialize generator
generator = CommentGenerator()

# Gradio Interface
with gr.Blocks(theme=gr.themes.Soft()) as app:
    # Hidden API key (set your actual key here)
    api_key = gr.Textbox(value="openrouter Ai key", visible=False)

    with gr.Row():
        with gr.Column(scale=3):
            gr.Markdown("""
            ## 🏥 Healthcare AI Thought Leader Comments
            *Generate professional comments with integrated emojis*
            """)

            post_content = gr.Textbox(
                label="📄 Paste a healthcare AI LinkedIn post",
                lines=5,
                placeholder="Paste the post content here..."
            )

            client_bio = gr.Textbox(
                label="👔 Your professional bio/expertise",
                lines=3,
                placeholder="E.g.: 'AI solutions at Hospital X, specializing in...'"
            )

            generate_btn = gr.Button("✨ Generate Comments (2-3 variants)", variant="primary")

            output1 = gr.Textbox(
                label="💬 Comment Option 1",
                lines=3,
                interactive=True
            )
            output2 = gr.Textbox(
                label="💬 Comment Option 2",
                lines=3,
                interactive=True,
                visible=False
            )
            output3 = gr.Textbox(
                label="💬 Comment Option 3",
                lines=3,
                interactive=True,
                visible=False
            )

        with gr.Column(scale=2):
            history_btn = gr.Button("📚 Show History", variant="secondary")
            history_output = gr.HTML(
                label="Comment History",
                visible=False
            )

    # Format history as HTML for better display
    def format_history():
        if not generator.history:
            return "<div style='padding: 20px; color: #666'>No history yet</div>"

        history_html = "<div style='max-height: 500px; overflow-y: auto; padding: 10px'>"
        for entry in reversed(generator.history):
            history_html += f"""
            <div style='border-bottom: 1px solid #eee; padding: 15px 0; margin-bottom: 10px'>
                <div style='font-size: 0.8em; color: #888; margin-bottom: 5px'>
                    {entry['time']} • {entry['post']}
                </div>
                <div style='font-size: 1em; line-height: 1.4'>
                    {entry['comment']}
                </div>
            </div>
            """
        history_html += "</div>"
        return history_html

    # Event handlers
    def update_outputs(comments):
        outputs = [gr.update(value="", visible=False) for _ in range(3)]
        for i, comment in enumerate(comments):
            if i < 3:
                outputs[i] = gr.update(value=comment, visible=True)
        return outputs

    generate_btn.click(
        fn=lambda p, b: generator.generate_comments_with_emojis(p, b, api_key.value),
        inputs=[post_content, client_bio],
        outputs=[]
    ).then(
        fn=lambda: update_outputs(generator.current_comments),
        outputs=[output1, output2, output3]
    )

    history_btn.click(
        fn=lambda: gr.update(visible=not history_output.visible),
        outputs=[history_output]
    ).then(
        fn=format_history,
        outputs=[history_output]
    )

app.launch(share=True)
```
